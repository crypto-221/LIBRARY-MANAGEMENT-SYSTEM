## `LIBRARY MANAGEMENT SYSTEM`

```python
available_books = ['Python learning' , 'deep learning' , 'data science' , 'jhon elia' , 'machine learning' ,'can we be stranger again' ]
issued_books = []
def AddBook():
    book_name = input("Enter book name here you want to add:-")
    available_books.append(book_name)
    print("Successfully added to Book list!!!")
def IssueBook():
    issue_book = input("Enter book name here you want to issue from book library:-")
    if issue_book in available_books:
        issued_books.append(issue_book)
        print("Successfully issued from library")
        available_books.remove(issue_book)
    else:
        print("Soryy we cant provide you this book , its not available in library!!")    
def ReturnBook():
    return_book = input("Enter your book name here you want to return:-")
    if return_book in issued_books :
        available_books.append(return_book)
        print("Successfully return!!")
    else:
        print("Soryy!! right now this book is not in issued book so you cant return it..")

def DisplayBooks():
    i = 0
    while i < len(available_books):
        print(f"{i}. {available_books[i]}")
        i+=1

def Menu():
    print("LIBRARY MANAGEMENT SYSTEM ")
    while True:
        print("1.AddBook\n2.IssueBook\n3.ReturnBook\n4.DisplayBooks\n5.Exit")
        user_choice = int(input("Enter your choice here:-"))      
        if user_choice == 1 :
            AddBook()
            print("-"*40 )
        elif user_choice == 2 :
            IssueBook()
            print("-" *40 )
        elif user_choice == 3 :
            ReturnBook()
            print("-"*40 )
        elif user_choice == 4 :
            DisplayBooks()
            print("-"*40 )
        elif user_choice == 5 :
            print("Successfully exiting from library...")
            print("-"*40 )
            break
        else:
            print("Soryy!! wrong choice try again later...")
Menu()
```
